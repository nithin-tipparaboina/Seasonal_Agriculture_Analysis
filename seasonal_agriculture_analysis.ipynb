## 3. Data Cleaning

Missing values were identified in the dataset and handled using the median value of the respective columns. This approach helped retain all 4,000 records while preparing the data for analysis.

```python
df["Rainfall_mm"] = df["Rainfall_mm"].fillna(df["Rainfall_mm"].median())

df["Soil_Moisture_pct"] = df["Soil_Moisture_pct"].fillna(
    df["Soil_Moisture_pct"].median()
)

df["Yield_Tonnes_Ha"] = df["Yield_Tonnes_Ha"].fillna(
    df["Yield_Tonnes_Ha"].median()
)

print("Missing values handled successfully!")
```

## Key Findings

### Seasonal Performance
- Kharif season recorded the highest average yield of 5.63 tonnes/ha.
- Kharif also had the highest average profit of approximately ₹1,78,915.
- Zaid recorded the lowest average profit, with an average loss of approximately ₹24,805.
- Kharif had the highest average production and water efficiency among the three seasons.
- Kharif also showed the highest average disease and pest risk.

### Irrigation Performance
- Drip irrigation recorded the highest average yield at 6.58 tonnes/ha.
- Drip irrigation also recorded the highest average profit of approximately ₹2,19,626.
- Rainfed farming recorded the highest water efficiency at 7.56 tonnes per 1000 m³.
- This shows a trade-off between higher yield/profit and water efficiency.

### Crop Performance
- Sugarcane recorded the highest average yield and average profit.
- Sugarcane also had the highest average water usage.
- Chilli recorded the second-highest average profit.
- Maize, Rice and Wheat showed negative average profits in this dataset.
- Sugarcane showed the highest average water efficiency among the crops in the dataset.

### Geographical Performance
- Punjab recorded the highest average crop yield and average profit among the states.
- Andhra Pradesh recorded the lowest average yield and average profit among the states.

### Environmental and Farm Factors
- Rainfall and temperature showed very weak linear relationships with crop yield.
- Water usage showed a moderate positive relationship with crop yield.
- Crop yield showed considerable variation across nitrogen, fertilizer and seed quality levels.
- Higher pesticide usage was generally associated with lower observed disease and pest risk.
- Higher crop yield was generally associated with higher profit.

## Conclusion

This analysis examined seasonal agricultural performance using yield, production, profit, water usage, water efficiency, environmental conditions, irrigation methods, crop types and geographical factors.

The analysis shows that Kharif performed better than Rabi and Zaid in terms of average yield, production, profit and water efficiency. Drip irrigation recorded the highest average yield and profit, while Rainfed farming recorded the highest water efficiency among irrigation methods.

Crop-level analysis showed significant differences in yield, profitability and water usage. Sugarcane recorded the highest average yield and profit, but also required the highest average water usage. Geographical analysis showed that Punjab had the highest average yield and profit among the states in the dataset.

The relationship analysis also showed that rainfall and temperature had very weak linear relationships with yield, while water usage had a moderate positive relationship with yield. Overall, agricultural performance depends on multiple factors rather than a single environmental or farming factor.

The findings can help farmers and agricultural decision-makers compare seasonal performance, irrigation practices, crop profitability and resource usage to support more informed agricultural planning.

## Recommendations

1. Farmers can consider suitable irrigation methods such as drip irrigation where higher yield and profitability are the priority.

2. Water-efficient practices should be encouraged, especially in areas where water availability is limited.

3. Crop selection should consider both expected profitability and water requirements rather than yield alone.

4. Seasonal planning can focus on the stronger performance observed during the Kharif season while considering its higher disease and pest risk.

5. Farmers should monitor disease and pest conditions and use crop protection practices according to observed risk levels.

6. Agricultural decisions should consider multiple factors such as soil conditions, water availability, irrigation method, crop type, seasonal conditions and expected profitability.

7. Further analysis with larger and more detailed datasets can help identify the combined effect of environmental and farming factors on agricultural performance.

## Limitations

- The dataset contains 4,000 farm records and represents the available observations in this analysis.
- Some values were missing in rainfall, soil moisture and yield columns and were filled using the median value.
- The analysis mainly uses averages, correlations and visualizations, so correlation should not be interpreted as causation.
- Different crops naturally have different yield scales, so yield alone should not be used to determine the best crop.
- The dataset does not provide enough information to establish causal relationships between farming practices and outcomes.
- Further analysis using larger datasets and advanced statistical or machine learning techniques could provide deeper insights.

## Final Summary

The analysis shows clear differences in agricultural performance across seasons, crops, irrigation methods and states. Kharif recorded the strongest overall seasonal performance, while Zaid showed lower average profitability.

Crop-level analysis revealed significant differences in profitability and resource usage. These findings provide useful data-driven insights for understanding seasonal agricultural performance and supporting better decisions related to crop selection, irrigation and resource management.
